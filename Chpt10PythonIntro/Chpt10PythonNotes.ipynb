# 10.8 Python Errors

## Compile (Translation)

Caught before the program runs. Also stops to make it not run.

## Exception (Run-Time Error)

The syntax is fine, so the program runs. However, becaus of this section, it will stop and give this error.

## Logic Error

Type of run-time error. The program runs normally and gives you output as it is supposed to. The problem, if you aren't paying attention, the output is wrong.

Logic Errors can be difficult to spot. This is why you have to have an understanding of what the output should be.

# 10.9 Algorithm Design

## Algorithm

A recipie for a solution

## Algorithms should be unambiguous

Precise instructions at every step, no guessing or personal opinion.

## Algorithms should be executable

Every step can actually be carried out in practice.

## Algorithms should be terminating

The steps eventually come to an end.
